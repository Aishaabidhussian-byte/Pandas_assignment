```python
import pandas as pd
data = {
    "Name": ["Ali", "Sara", "Ahmed","Ayessha","Zain"],
    "Marks": [70, 85, 60,95,45],
}

df = pd.DataFrame(data)
df
```

```python
Marks = df["Marks"]
print("\nAverage Marks:")
print(Marks.mean())
print("\nMaximum Marks:")
print(Marks.max())
print("\nMinimum Marks:")
print(Marks.min())
```

```python
Name = df["Name"]
print("\nTotal student:")
print(Name.count())
```

```python
passed = df[df["Marks"] >= 50]
print("\nPassed Students:")
print(passed)

Failed = df[df["Marks"] <= 50]
print("\nFailed Students:")
print(Failed)
```
